```sql
---create database sparksample_DB ;
use sparksample_DB;
create table customers(
    Customerid int,
    name   string,
    country string
)
USING
PARQUET;
```

```sql
INSERT INTO customers
values
    (1,'Raj','UK');
```

```sql
update customers set country = 'USA'
```

```sql
create table customers_delta(
    Customerid int,
    name   string,
    country string
)
USING
DELTA;
```

```sql
INSERT INTO customers_delta
values
    (1,'Raj','UK');
```

```sql
update customers_delta set country = 'USA'
```

```sql
--- Time travel : you can go back to the previous version
describe history customers_delta
```

```sql
select * from customers_delta version as of 1;
```

```sql
--- clean the unused version files : House keeping 
vacuum customers_delta;

---- by default : it will delete only files that are older than 7 days (168 hrs)retenetion period
```

```sql
---- force vacuum
--SET spark.databricks.delta.retentionDurationCheck.enabled = true;
vacuum customers_delta  retain 300 hours;
```

```sql
---after vacuum you cannot time travel 
select * from customers_delta version as of 1;
```

```sql
select * from customers_delta
```

```sql
update customers_delta set country = 'India'
```

```sql
describe history customers_delta
```

```sql
select * from customers_delta 
```

```sql
restore customers_delta to version as of 8;
```
