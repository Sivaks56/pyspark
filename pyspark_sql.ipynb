```python
#creating dataframe
data = [
    (1,"Raj","Chennai","23000"),
    (2,"Ravi","Mumbai","25000"),
    (3,"Rakesh","Ooty","29000")
]

columns = ["id","name","city","salary"]

df = spark.createDataFrame(data,columns)

df.show()
```

```python
df.printSchema()
```

```python
df.columns
```

```python
df.count()
```

```python
df.describe().show()
```

```python
#sql vs pyspark
# select
df.select("id","name").show()
```

```python
#select - 2
df.select(df.name , df.id).show()
```

```python
#filter 
#sql --- where
df.filter(df.salary > 25000).show()
```

```python
#filyter -- using where
df.where(df.salary > 25000).show()
```

```python
#filtering using multiple conditions: and --- & or --- |
df.filter(
    (df.salary > 24000) | (df.city == 'Ooty')
    ).show()
```

```python
from pyspark.sql.functions import *
# col --- column
df.filter(
    (col("salary") > 24000) | (col("city") == 'Ooty')
    ).show()
```

```python
# filtring using in
df.filter(
    col("city").isin("Chennai","Mumbai")
).show()
```

```python
#not in
df.filter(
    ~col("city").isin("Chennai","Mumbai")
).show()
```

```python
#distinct
df.select("city").distinct().show()
```

```python
#order by : asc
df.orderBy("salary").show()
```

```python
#order by : desc
df.orderBy(col("salary").desc()).show()
```

```python
#limit 
df.limit(1).show()
```

```python
#rename column in pyspark
df.withColumnRenamed(
    "name",
    "Employee_Name"
).show()
```

```python
#how to add new column --
df.withColumn(
    "annual_salary",
    col("salary")*12
).show()
```

```python
#groupby
df.groupBy(col("city")).agg(count("id")).show()
```

```python
#alias
df.groupBy(col("city")).agg(count("id").alias("employee count")).show()
```

```python
#group by + having in pyspark
df.groupBy(col("city")).agg(avg("salary").alias("average salary")).filter(col("average salary") > 23000).show()
```

```python
employee_data = [
    (101, "Ravi", 10, 50000),
    (102, "Priya", 20, 60000),
    (103, "Arun", 10, 55000),
    (104, "Meena", 30, 70000),
    (105, "Karthik", 40, 65000)
]

employee_columns = ["emp_id", "emp_name", "dept_id", "salary"]

df_emp = spark.createDataFrame(employee_data, employee_columns)

df_emp.show()
```

```python
dept_data = [
    (10, "IT"),
    (20, "HR"),
    (30, "Finance"),
    (50, "Marketing")
]

dept_columns = ["dept_id", "dept_name"]

df_dept = spark.createDataFrame(dept_data, dept_columns)

df_dept.show()
```

```python
#join in pyspark
df_joined = df_emp.join(df_dept,
df_emp.dept_id == df_dept.dept_id,
"inner"
).select(
    df_emp.emp_name,
    df_emp.dept_id,
    df_dept.dept_name
).show()

#df.join(second df , condituon , join type)
```

```python
df_joined = df_emp.join(df_dept,
df_emp.dept_id == df_dept.dept_id,
"right"
).select(
    df_emp.emp_name,
    df_emp.dept_id,
    df_dept.dept_name
).show()
```
