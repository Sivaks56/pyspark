```python
#CREATING dataframe
data = [
    (101, "Arun",   "IT",      60000, "Chennai",   2),
    (102, "Priya",  "HR",      50000, "Bangalore", 1),
    (103, "Kumar",  "IT",      75000, "Chennai",   3),
    (104, "Divya",  "Finance", 65000, "Chennai",   2),
    (105, "Ravi",   "HR",      55000, "Bangalore", 1),
    (106, "Meena",  "IT",      80000, "Hyderabad", 4),
    (107, "Suresh", "Finance", 70000, "Chennai",   3),
    (108, "Anu",    "IT",      62000, "Bangalore", 2),
    (109, "Vijay",  "HR",      48000, "Chennai",   1),
    (110, "Lakshmi","Finance", 72000, "Hyderabad", 3)
]

columns = [
    "EmpID",
    "EmpName",
    "Department",
    "Salary",
    "City",
    "Experience"
]

df = spark.createDataFrame(data,columns)
df.show()

```

```python
df.printSchema() # it will give automatic schema info
```

```python
df.columns # it is method that's why there is no bracket
```

```python
df.count() # to find how much records are there
```

```python
df.describe().show() # it will show methomatical information
```

```python
#sql vs pyspark
#select which columns we want
df.select("EmpID","EmpName").show()

```

```python
# secound type select
df.select(df.EmpName,df.EmpID).show()
```

```python
#filter
#sql --- where
df.filter(df.Salary > 60000).show()
```

```python
#filter using where
df.where(df.Salary > 60000).show()
```

```python
#filtering using multiple condition
df.filter((df.Salary > 60000) & (df.City == 'Chennai')).show()
```

```python
# or condition using pipe simple
df.filter((df.Salary > 60000) | (df.City == 'Chennai')).show()
```

```python
from pyspark.sql.functions import *
#col --- column
df.filter((col("Salary") > 60000) & (col("City") == 'Chennai')).show()
```

```python
#filter using in like sql in condition
df.filter(col("City").isin("Chennai","Bangalore")).show()
```

```python
#not in  city la "Chennai","Bangalore" illama mithi irukura city varum
df.filter(
    ~col("City").isin("Chennai","Bangalore")
).show()
```

```python
#distinct it will remove duplicate value give unique value
df.select("City").distinct().show()
```

```python
#order by : asc
df.orderBy("Salary").show()
```

```python
#orderby desc
df.orderBy(col("Salary").desc()).show()
```

```python
#limit it will give first 4 rows
df.limit(4).show()
```

```python
#remane column in pyspark
df.withColumnRenamed(
    "EmpName",
    "Emp_Name"

).show()

```

```python
#how to add new column
df.withColumn("Annual_Salary",col("Salary")*12).show()
```

```python
#groupby
df.groupBy(col("City")).agg(count("EmpID")).show()
```

```python
#alies
df.groupBy(col("City")).agg(count("EmpID").alias("Emp_count")).show()
```

```python
#having in pyspark
df.groupBy(col("City")).agg(avg("Salary").alias("avg_salary")).filter(col("avg_salary") > 23000).show()
df.groupby(col("city").agg(avg("salary").alias("avg_salary")).filter(col("avg_salary") > 23000)).show()
```

```python
dept_data = [
    (10, "IT", "Technology", "Chennai"),
    (20, "HR", "Human Resources", "Bangalore"),
    (30, "Finance", "Finance & Accounts", "Mumbai"),
    (40, "Sales", "Sales & Marketing", "Delhi"),
    (50, "Admin", "Administration", "Hyderabad")
]

dept_columns = [
    "DeptID",
    "Department",
    "DepartmentName",
    "DeptLocation"
]

df2 = spark.createDataFrame(
    dept_data,
    dept_columns
)

df2.show()
```

```python
# join in pyspark
df_joined = df.join(df2,df.Department == df2.Department,"inner").select(df.EmpID,df.EmpName,df2.Department).show()
```

```python
# join in pyspark
df_joined = df.join(df2,df.Department == df2.Department,"left").select(df.EmpID,df.EmpName,df2.Department).show()
```
