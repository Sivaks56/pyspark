```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName("JoinPractice").getOrCreate()

emp_data = [
    (1, "Ravi", 101, 50000),
    (2, "Priya", 102, 60000),
    (3, "Arun", 101, 45000),
    (4, "Divya", 103, 70000),
    (5, "Kumar", None, 40000),
    (6, "Sneha", 105, 65000)
]

emp_columns = ["emp_id", "emp_name", "dept_id", "salary"]

df1 = spark.createDataFrame(emp_data, emp_columns)

df1.show()


dept_data = [
    (101, "IT"),
    (102, "HR"),
    (104, "Finance"),
    (106, "Marketing"),
    (None, "Unknown")
]

dept_columns = ["dept_id", "dept_name"]

df2 = spark.createDataFrame(dept_data, dept_columns)

df2.show()
```

```python
IT_df = df1.alias("e").join(
    df2.alias("d"),
    col("e.dept_id") == col("d.dept_id"),
    "inner"
).select(
    col("e.emp_id"),
    col("e.emp_name"),
    col("d.dept_id"),
    col("d.dept_name"),
    col("e.salary")
).filter(col("d.dept_name") == "IT")




hr_df = df1.alias("e").join(
    df2.alias("d"),
    col("e.dept_id") == col("d.dept_id"),
    "inner"
).select(
    col("e.emp_id"),
    col("e.emp_name"),
    col("d.dept_id"),
    col("d.dept_name"),
    col("e.salary")
).filter(col("d.dept_name") == "HR")


IT_df.unionByName(hr_df).show() # it will join top bottom
```

```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import *

data = [
    (1, "ravi kumar",  "IT",      "60000", "2022-01-15"),
    (2, "PRIYA SHARMA","HR",      "45000", "2021-06-10"),
    (3, "arun kumar",  "IT",      "75000", "2020-03-20"),
    (4, "Divya",       "Finance", "55000", "2023-08-01"),
    (5, "sneha",       "IT",      "80000", "2019-11-25")
]

columns = ["emp_id", "name", "department", "salary", "join_date"]

df = spark.createDataFrame(data, columns)

df.show()
df.printSchema()
```

```python
from pyspark.sql.functions import col, when

df = df.withColumn(
    "salary_grade",
    when(
        (col("department") == "IT") & (col("salary") > 70000),
        "senior IT"
    ).when(
        (col("department") == "IT") & (col("salary") > 55000),
        "Mid level IT"
    ).otherwise("low")
)

df.show()

```

```python
df = df.withColumn("country", lit("India"))

df = df.withColumn(
    "bonus",
    col("salary").cast("int") * lit(0.10)
)
df.show()
```

```python
df.select(
    "name",
    year(to_date("join_date")).alias("year"),
    month(to_date("join_date")).alias("month"),
    dayofmonth(to_date("join_date")).alias("day")
).show()
```

```python
from pyspark.sql.functions import *
df.select("name",
          datediff(current_date(), to_date("join_date")).alias("days_worked")).show()

df = df.withColumn(
    "after_30_days",
    date_add(to_date("join_date"), 30)
)

df.show()
```

```python
df.filter(
    year(to_date("join_date")) > 2022
).show()
```

```python
df = df.withColumn(
    "experience",
    months_between(current_date(), to_date(col("join_date")))
)


df = df.withColumn("experiance_year", 
                   floor(months_between(current_date(), to_date("join_date"))/12))

df.show()
```

```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import *
from pyspark.sql.window import Window


data = [
    (1, "Ravi",   "IT", 80000),
    (2, "Priya",  "IT", 80000),
    (3, "Arun",   "IT", 60000),
    (4, "Divya",  "HR", 50000),
    (5, "Sneha",  "HR", 70000),
    (6, "Kumar",  "HR", 40000)
]

columns = ["emp_id", "name", "department", "salary"]

df = spark.createDataFrame(data, columns)
df.show()
```

```python
window_spec = Window.partitionBy("department").orderBy(
    col("salary").desc()
)

df.withColumn(
    "row_num",
    row_number().over(window_spec)
).show()
```

```python
window_spec = Window.partitionBy("department").orderBy(
    col("salary").desc()
)

df.withColumn(
    "rank",
    dense_rank().over(window_spec)
).show()

df.withColumn(
    "rank",
    rank().over(window_spec)
).show()


PTl = window.partitionby("department").orderby(col("salary").desc())
df.withcolumn("rank",rank().over(ptl)).show()

```

```python
window_spec = Window.partitionBy("department").orderBy("emp_id")

df.withColumn(
    "previous_salary",
    lag("salary", 1).over(window_spec)
).show()

window_spec = Window.partitionBy("department").orderBy("emp_id")

df.withColumn(
    "Next_salary",
    lead("salary", 1).over(window_spec)
).show()
```
