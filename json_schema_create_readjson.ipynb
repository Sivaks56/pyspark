```python
from pyspark.sql.functions import *
from pyspark.sql.types import *

data = [("babu",30,"mumbai"),("manoj",28,"chennai")]

schema = StructType([StructField("name",StringType(),True),
                     StructField("age",IntegerType(),True),
                     StructField("City",StringType(),True)])

df = spark.createDataFrame(data,schema)
df.show()
```

```python
#define schema for nested json
from pyspark.sql.functions import *
from pyspark.sql.types import *

data1 = [(1,{"name":"babu","age":30,"City":"mumbai"}),
         (2,{"name":"manoj","age":28,"City":"chennai"})]

schema1 = StructType([StructField("id",IntegerType(),True),
                     StructField("data",StructType([StructField("name",StringType(),True),
                                                    StructField("age",IntegerType(),True),
                                                    StructField("City",StringType(),True)]),True)])
                     
df2 = spark.createDataFrame(data1,schema1)
df2.show()
```

```python
df2.select("id","data.*").show() #read json file
```
