```python
from pyspark.sql import *
from pyspark.sql.types import *
data = [("vija quick",),("vignesan Saravanan",),("Manoj Sundar",),("Arun Babu",)]
df = spark.createDataFrame(data,["FullName"])
display(df)
```

```python
from pyspark.sql.functions import *

df2 = (
    df.withColumn("FirstName", split(col("FullName"), " ").getItem(0))
      .withColumn("LastName", split(col("FullName"), " ").getItem(1))
)

display(df2)
```

```python
from pyspark.sql import *
from pyspark.sql.types import *
Data = [(1,"Alice",None,4000),
        (2,None,2000,None),
        (3,"Charlie",3000,7000),
        (4,None,None,6000),
        (None,None,None,None),
        (3,"Charlie",3000,7000),
        (5,"hari",4000,5000),
        (6,"hari",5000,6000)]

schema = ["id","name","salary","yearsales"]

df3 = spark.createDataFrame(Data,schema)
display(df3)
```

```python
df3 = df3.fillna(0, subset=["salary"]) #subset kodutha tha particuler column fill pannum kodukalana all column fill pannidum
display(df3)
```

```python
df3 = df3.fillna(value=2000,subset=["salary"]) # value = 2000 it will give default value to particuler column
display(df3)
```

```python
df3 = (
    df3.fillna(value=2000, subset=["salary"])
       .fillna(value=5000, subset=["yearsales"])
)

display(df3)
```

```python
df5 = df3.dropna(how="all",subset=["name"]) #name column la enga lam null iruko ellam delete aakidum
display(df5)
```

```python
df3 = df3.dropna(how="any",subset=["id"])
display(df3)
```

```python
df3 = df3.dropna(how="all") #it will delete entire rows has null value remove that row
display(df3)
```

```python
df3 =df3.fillna("unknown",subset=["name"])

display(df3)
```

```python
df3 = df3.dropDuplicates()
display(df3)
```

```python
df3 = df3.dropDuplicates(["name"])

display(df3)
```
