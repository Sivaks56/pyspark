```python
data = [
    ("B001", "Arjun", "2024-01-05 10:15:00", "2024-01-05 13:45:00"),
    ("B002", "Meena", "2024-01-07 09:30:00", "2024-01-07 12:00:00"),
    ("B003", "Ravi", "2024-01-10 15:00:00", "2024-01-11 10:00:00"),
    ("B004", "Suma", "2024-01-12 18:45:00", "2024-01-13 08:15:00"),
    ("B005", "Karan", "2024-01-15 08:00:00", "2024-01-15 09:30:00")
]

df = spark.createDataFrame(data, ["OrderID", "CustomerName", "OrderTime", "DeliveryTime"])
display(df)
```

```python
#it will change date format
from pyspark.sql.functions import *

df = df.withColumn("OrderTime", to_date(col("OrderTime"))) \
       .withColumn("DeliveryTime", to_date(col("DeliveryTime")))

display(df)
```

```python
#it will change date format
from pyspark.sql.functions import *
df = df.withColumn("OrderTime", to_date(col("OrderTime"))).withColumn("DeliveryTime", to_date(col("DeliveryTime")))

display(df)
```

```python
#it will change to timestamp
df4 =df.withColumn("OrderTime", to_timestamp("OrderTime","yyyy-MM-dd HH mm ss SSS"))
display(df4)
```

```python
#split timestamp to date hour minute second millisecond
df4 =df.withColumn("OrderTime", to_timestamp("OrderTime","yyyy-MM-dd HH:mm:ss"))\
    .withColumn("date",date_format("OrderTime","yyyy-MM-dd"))\
    .withColumn("hour",date_format("OrderTime","HH"))\
    .withColumn("minute",date_format("OrderTime","mm"))\
    .withColumn("second",date_format("OrderTime","ss"))
display(df4)
```

```python
#split day month year dayname
df5 = (df.withColumn("day_name", date_format("OrderTime", "E"))
    .withColumn("day", date_format("OrderTime", "dd"))
    .withColumn("Month", date_format("OrderTime", "MM"))
    .withColumn("Year", date_format("OrderTime", "yyyy")))

display(df5)
```
