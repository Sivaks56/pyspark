```python
data = [
    (101, "Arun", "IT", 60000, 28, "Chennai"),
    (102, "Priya", "HR", 50000, 32, "Bangalore"),
    (103, "Rahul", "IT", 75000, 35, "Chennai"),
    (104, "Divya", "Finance", 65000, 29, "Mumbai"),
    (105, "Karthik", "IT", 55000, 25, "Hyderabad"),
    (106, "Sneha", "HR", 58000, 27, "Chennai"),
    (107, "Vijay", "Finance", 72000, 38, "Bangalore"),
    (108, "Meena", "IT", 80000, 31, "Mumbai"),
    (109, "Ajay", "Finance", 60000, 26, "Chennai"),
    (110, "Anitha", "HR", 52000, 30, "Hyderabad")
]

columns = ["emp_id", "name", "department", "salary", "age", "city"]

df = spark.createDataFrame(data, columns)

df.show()
```

```python
#1. Display Specific Columns
df.select("emp_id","name","department","salary").show()
```

```python
#2. Filter Employees
#Salary greater than 60,000.
df.filter(df.salary > 60000).select("emp_id","name","department","salary").show()
```

```python
#3. Multiple Filter Conditions
#Find employees who belong to the IT department AND have salary greater than 60,000.
from pyspark.sql.functions import *
df.filter((df.department == "IT") & (df.salary > 60000)).select(col("emp_id"),col("name"),col("department"),col("salary")).show()
```

```python
#4. Create a New Column
#Create a column called annual_salary using salary × 12. Display name, salary and annual_salary.

df.withColumn("annaul_salary",df.salary*12).select("name","salary","annaul_salary").show()
```

```python
#5. Department-wise Employee Count
#Find the number of employees in each department. Use groupBy() and count().
df.groupBy("department").count().show()
```

```python
#6. Department-wise Salary Analysis
#For each department, calculate average salary, maximum salary and minimum salary using groupBy() and
#agg().
df.groupBy("department").agg(avg("salary").alias("avg_salary"),max("salary").alias("max_salary"),min("salary").alias("min_salary")).show()
```

```python
#7. Sort Employees
#Display all employees in descending order of salary.
df.sort(df.salary.desc()).show()
```

```python
#8. Multiple Filter Conditions
#Salary greater than 70,000 OR age less than 27.
#df.filter((df.salary > 70000) | (df.age < 27)).show()
df.filter((col("salary") > 70000) | (col("age") < 27)).show()
```

```python
#8. Filter Using Multiple Conditions
#Find employees whose salary is greater than 70,000 OR whose age is less than 27.
df.filter((col("salary") > 70000) | (col("age") < 27)).show()
```

```python
#9. Highest Salary by Department
#Find the maximum salary for each department. Bonus: display the employee name along with the maximum
#salary.
max_salary_df = df.groupBy("department").agg(max("salary").alias("max_salary"))

result = df.alias("e").join(
    max_salary_df.alias("m"),
    (col('e.department') == col('m.department')) &
    (col('salary') == col('max_salary'))
)

result.select(
    col('e.department'),
    col('e.name'),
    col('e.salary')
).show()
```

```python
# 10. Employee Salary Category
# Create a column called salary_category using these rules:
# salary >= 70,000 fi High
# salary >= 55,000 fi Medium
# salary < 55,000 fi Low
# Display name, salary and salary_category.
df.withColumn("salary_category",when(df.salary >= 70000, "High").when(df.salary >= 55000,"medium").when(df.salary < 55000,"Low")).select("name", "salary", "salary_category").show()
```
