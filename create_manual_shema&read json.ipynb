```python
#manual schema create keyword struct type and struct field
from pyspark.sql.types import *
schema = StructType([
    StructField("emp_id", LongType()),
    StructField("emp_name", StringType()),
    StructField("city", StringType()),
    StructField("department", StringType()),
    StructField("salary", LongType()),
    StructField("sale_date", StringType()),
    StructField("order_id", LongType()),
    StructField("product", StringType()),
    StructField("quantity", LongType()),
    StructField("sales_amount", LongType())
])

df = spark.createDataFrame(data,schema) #created schema give inside dataframe
df.show()
df.printSchema()
```
